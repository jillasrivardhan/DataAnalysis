```python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("Data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df = df.drop(columns=['Age','Gender'])
```

```python
X = df.drop(columns=['Salary'])
```

```python
y = df['Salary']
```

```python
num_X = ['Experience','Performance_Score','Years_at_Company']
cat_X = ['City','Education','Department','Job_Level']
```

```python
from sklearn import pipeline
cat = Pipeline(steps=[
    ('imputer',SimpleImputer(strategy='most_frequent')), #simpleimputer fills null values with mostfrequent
    ('onehot',OneHotEncoder(dtype=int)) # one hot encoder turn categorical data into numerical data
])
```

```python
num = Pipeline(steps=[
    ('imputer',SimpleImputer(strategy='mean')),
    ('scaler',StandardScaler())
])
```

```python
preprocessor = ColumnTransformer(transformers=[('numeric',num,num_X),("categoric",cat,cat_X)])
```

```python
model = Pipeline(steps=[('preprocessor',preprocessor),
                        ('classifier',LinearRegression())])
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
mse = mean_squared_error(y_test,y_pred)
```

```python
mse
```

```python
# rmse = np.sqrt(mse)
rmse = mse ** 0.5
```
