## LABLE ENCODER

```python
#lable encoder is used to encode the values into unit values.
#This is used only for ordinal data.
#ordinal data means the values are related to each other(having relation).

le = LabelEncoder()
```

```python
scd = ['Gender','Education','Job_Level','Department','City']
```

```python
for col in scd:
  df[col] = le.fit_transform(df[col])
```
